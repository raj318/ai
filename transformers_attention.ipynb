# Learn transformers from scratch

```python
# self attention

# Query : what am i looking for

# Key : what information do i contain?

# Value : What information should i actually provide? 

# Q = X * Wq
# K = X * Wk
# V = X * Wv

import torch
import math

torch.manual_seed(42) 

X = torch.tensor([
    [0.2, 0.4, 0.1, 0.7],
    [0.8, 0.1, 0.6, 0.2],
    [0.3, 0.9, 0.2, 0.5],
    [0.7, 0.2, 0.8, 0.1]
])

print(X.shape)
```

```python
d_model = 4
d_k = 4

W_q = torch.randn(d_model, d_k)
W_k = torch.randn(d_model, d_k)
W_v = torch.randn(d_model, d_k)

print(W_q)
print(W_k)
print(W_v)
```

```python
Q = X @ W_q
K = X @ W_k
V = X @ W_v

print(Q)
print('\n')

print(K)
print('\n')

print(V)
print('\n')
```

```python
scores = Q @ K.T # have to multiply with each other word k value that is why we need it to transpose
print(scores)
print(scores.shape)
```

```python
scores = scores/math.sqrt(d_k)
print(scores)
```

```python
attention_weights = torch.softmax(scores, dim=-1)
print(attention_weights)
```

```python
output = attention_weights @ V
print(output)
print(output.shape)
```

# multi headed self attention

```python
seq_len = 4
d_model = 8
num_heads = 2

d_k = d_model // num_heads

X = torch.randn(seq_len, d_model)

print(f"x = {X.shape}")
print(f"d_k = {d_k}")
```

```python
W_Q = torch.randn(d_model, d_model) # d_k before splitting based on head is same as d_model, we progress which is better to make the model learn the parameters
W_K = torch.randn(d_model, d_model)
W_V = torch.randn(d_model, d_model)

Q = X @ W_Q

K = X @ W_K

V = X @ W_V

print(Q.shape)
print(K.shape)
print(V.shape)
print('\n')
print(Q)
```

```python
Q = Q.view(seq_len, num_heads, d_k) # 4, 2, 4

print(Q)

Q = Q.transpose(0, 1)

print(Q.shape)

print('\n')

print(Q)
```

```python
K = K.view(seq_len, num_heads, d_k).transpose(0, 1)
V = V.view(seq_len, num_heads, d_k).transpose(0, 1)

print(K.shape)
print(V.shape)
```

```python
scores = Q @ K.transpose(-2, -1)
print(scores.shape)
```

```python
scores = scores / math.sqrt(d_k)

attention_weights = torch.softmax(scores, -1)

output = attention_weights @ V

print(output.shape)

W_O = torch.randn(d_model, d_model)

print(W_O)

```

```python
head_outputs = output.transpose(0, 1)

print(head_outputs.shape)

head_output = head_outputs.reshape(seq_len, d_model) # 4, 8

print(head_output.shape)

print(W_O.shape)

output = head_output @ W_O

print(output.shape)
```

# Trainable pytorch Layer

```python
import torch.nn as nn

class MultiHeadAttention(nn.Module):
    """Class for multi head attention layer for transfomers"""

    def __init__(self, d_model, num_heads):

        super().__init__()

        self.model = d_model
        self.num_heads = num_heads

        self.d_k = int(self.model/self.num_heads)

        self.W_Q = nn.Parameter(
            torch.randn(d_model, d_model) # don't want to consider d_k yet as there will head split for attention
        )

        self.W_K = nn.Parameter(
            torch.randn(d_model, d_model)
        )

        self.W_V = nn.Parameter(
            torch.randn(d_model, d_model)
        )

        self.W_O = nn.Parameter(
            torch.randn(d_model, d_model)
        )

    def forward(self, X):
        Q = X @ self.W_Q
        K = X @ self.W_K
        V = X @ self.W_V

        Q = Q.reshape(-1, self.num_heads, self.d_k).transpose(0,1) # -1 calcualtes dimention automatically based on remaining dimensions
        K = K.reshape(-1, self.num_heads, self.d_k).transpose(0,1)
        V = V.reshape(-1, self.num_heads, self.d_k).transpose(0,1)

        scores = Q @ K.transpose(-2, -1)
        scores = scores/math.sqrt(self.d_k)
        attention_weights = torch.softmax(scores, dim=-1)

        heads_output = attention_weights @ V  # heads output shape = (2, 4, 4) 

        print(heads_output.shape)
        heads_output = heads_output.transpose(0, 1) # it became 4, 2, 4

        head_output = heads_output.reshape(-1, self.model) # 4, 8 -> combined two heads here

        output = head_output @ self.W_O

        return output        
```

```python
X = torch.randn(4, 8)

attention = MultiHeadAttention(
    d_model = 8,
    num_heads = 2
)

output = attention(X)

print(X.shape)
print(output.shape)
```

```python
for name, parameter in attention.named_parameters():
    print(name, parameter.requires_grad, parameter.shape)
```

```python
loss = output.sum()
loss.backward()
print(attention.W_Q.grad)
print(loss)
```

```python
import torch.nn as nn

class MultiHeadAttention(nn.Module):
    """Class for multi head attention layer for transfomers with batch size"""

    def __init__(self, d_model, num_heads):

        super().__init__()

        self.model = d_model
        self.num_heads = num_heads

        self.d_k = int(self.model/self.num_heads)

        self.W_Q = nn.Parameter(
            torch.randn(d_model, d_model) # don't want to consider d_k yet as there will head split for attention
        )

        self.W_K = nn.Parameter(
            torch.randn(d_model, d_model)
        )

        self.W_V = nn.Parameter(
            torch.randn(d_model, d_model)
        )

        self.W_O = nn.Parameter(
            torch.randn(d_model, d_model)
        )

    def forward(self, X):
        """ forward actual code"""
        Q = X @ self.W_Q
        K = X @ self.W_K
        V = X @ self.W_V

        Q = Q.reshape(-1, self.num_heads, self.d_k).transpose(0,1) # -1 calcualtes dimention automatically based on remaining dimensions
        K = K.reshape(-1, self.num_heads, self.d_k).transpose(0,1) # 4, 2, 4, after transpose it becomes 2, 4, 4
        V = V.reshape(-1, self.num_heads, self.d_k).transpose(0,1)

        scores = Q @ K.transpose(-2, -1)
        scores = scores/math.sqrt(self.d_k)
        attention_weights = torch.softmax(scores, dim=-1)

        heads_output = attention_weights @ V  # heads output shape = (2, 4, 4) 

        print(heads_output.shape)
        heads_output = heads_output.transpose(0, 1) # it became 4, 2, 4

        head_output = heads_output.reshape(-1, self.model) # 4, 8 -> combined two heads here

        output = head_output @ self.W_O

        return output        
```

```python
torch.manual_seed(42)

X = torch.randn(4, 8)

attention = MultiHeadAttention(
    d_model = 8,
    num_heads = 2
)

output = attention(X)

print(f"Input: {X.shape}")
print(f"Output: {output.shape}")
```

```python
for name, parameter in attention.named_parameters():
    print(
        name,
        parameter.shape,
        parameter.requires_grad # if this is true, pytorch will track grdients for the parameters
    )
```

```python
loss = output.sum()
loss.backward()
print(attention.W_Q.grad)
```
